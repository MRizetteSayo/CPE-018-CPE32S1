```python
from google.colab import files
uploaded = files.upload()

import cv2
from google.colab.patches import cv2_imshow

# Read an image
image = cv2.imread('orange.png')

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Display the image
cv2_imshow(image)

```

```python
whiter = image.copy()
whiter[whiter > 150] = 255
cv2_imshow(whiter)
```

```python
blacker = gray.copy()
blacker[blacker < 100] = 0
cv2_imshow(blacker)
```

```python
# ============ CENTER CROP ============
print("\n" + "="*50)
print("CENTER CROP (50%)")
print("="*50)

height, width = image.shape[:2]

# Calculate center crop dimensions (50% of original)
crop_width = width // 2
crop_height = height // 2

# Calculate starting point for center
start_x = (width - crop_width) // 2
start_y = (height - crop_height) // 2

# Crop the center
center_crop = image[start_y:start_y + crop_height, start_x:start_x + crop_width]

cv2_imshow(center_crop)
print(f"Cropped to: {crop_width}x{crop_height}")
print(f"Position: Top-left at ({start_x}, {start_y})")

# Save and download
cv2.imwrite('center_crop.jpg', center_crop)
files.download('center_crop.jpg')
```
